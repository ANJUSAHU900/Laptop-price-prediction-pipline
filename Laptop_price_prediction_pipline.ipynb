```python
import pandas as pd
df = pd.read_csv('/content/Cleaned_Laptop_data.csv')
df.head()
```

```python
# Reload the data to fix the NoneType error and then drop the columns
df = df.drop(columns=['display_size', 'warranty', 'msoffice'])
df.head()
```

```python
#-------------------------------------------------------
#RAM, SSD Aur HDD ko saaf karke number (integer) banana
#-------------------------------------------------------

# ram_gb se text hatakar integer banana (agar wo string type hai to)
if df['ram_gb'].dtype == 'object':
    df['ram_gb'] = df['ram_gb'].str.replace('GB GB','').astype(int)

# ssd se GB hatakar integer banana (agar wo string type hai to)
if df['ssd'].dtype == 'object':
    df['ssd'] = df['ssd'].str.replace('GB','').astype(int)

# hdd se GB htakar integer banana (agar wo string type hai to)
if df['hdd'].dtype == 'object':
    df['hdd'] = df['hdd'].str.replace('GB','').astype(int)

df.head()
```

```python
#jo gb change kiya use check kiya int m hua ya nhi

df.dtypes
```

```python
#-----------------------------------------------
#Missing (NULL) Values check karna aur clean krna
#-----------------------------------------------

#jo rows khali hai ya jisme price missing hai use drop karna h

df.dropna(subset= ['latest_price', 'brand'], inplace=True)
df.head()
```

```python
#--------------------------------------
# Discount column calculate karna
#--------------------------------------

#Discount amount = old_price - latest_price
df['discount_amount'] = df['old_price']-df['latest_price']

#Discount percentage
df['discount_percentage'] = round((df['discount_amount']/df['old_price'])*100,2)

#Check karna clean data ka sample
print("clean Data Shape:",df.shape)
df[['brand','model','ram_gb','ssd','latest_price','discount_percentage']].head()
```

```python
df.head()
```

```python
#If condition use kiya negative value hatane k liye
import numpy as np

#1.Discount Amount:
df['discount_amount'] = np.where(df['old_price']>0,df['latest_price'],0)

#discount percentage:
df['discount_percentage']= np.where(df['old_price']>0,round((df['discount_amount']/df['old_price'])*100,2),0)

#verify
df[['latest_price', 'old_price','discount_amount','discount_percentage']].head()
```

```python
#Display par % lagane k liye str lgayi(data number hi rahega)
df.head().style.format({'discount_percentage':'{:.2f}%'})
```

```python
#============================================================
#Q. Har brand ke total kitne laptops hain?
#soln. Brand column ka count nikal kar bar chart banaya.
#============================================================
# Matplotib library ko chart /graph banane ke liye'plt name se import kiya
import matplotlib.pyplot as plt

#1. Har brand ke total laptops count kiye aur dexcending order me sort kiya
brand_count = df['brand'].value_counts()
print(brand_count)

#Graph ke canvas ka size set kiya
plt.figure(figsize=(8,4))

#x-axis par brand names aur y-axis par unke count ka blue bar chart banaya
plt.bar(brand_count.index,brand_count.values,color= 'steelblue')

#Chart ke upar main title (heading)set kiya
plt.title("Total Laptops by Brand")

#Niche horizontal axis(x-axis) ka label 'Brand' diya
plt.xlabel("Brand")

# Side vertical axis (y-axis) ka label 'count' diya
plt.ylabel("Count")

# Brand names lambe hone par overlap na kare isliye text ko 45- degree ghuma diya
plt.xticks(rotation=45)

#Chart ke charo taraf ka faltu extra gap adjust kiya taki sb saaf dikhe
plt.tight_layout()

#final chart ko show krna
plt.show()
```

```python
from matplotlib.typing import LineStyleType
#====================================================
#Brand - wise Average Latest Price Analysis(Horizontal Bar Chart)
#========================================================

#Brand ke according group karke latest_price ka mean (average) nikala
brand_avg_price = df.groupby('brand')['latest_price'].mean().round(2)

# Graph mein sbse mehanga brand sbse upar dikhe isliye sort kiya
brand_avg_price = brand_avg_price.sort_values(ascending=True)

#Screen par values check krne k liye print kiya
print(brand_avg_price)

# Chart ka frame size set kiya (width:9, height: 5 inch)
plt.figure(figsize=(9,5))

# Horizontal bar chart (barh) banaya
plt.barh(brand_avg_price.index,brand_avg_price.values,color='#ff7f0e', edgecolor='black')

#visual ka title set kiya
plt.title("Average Laptop Price by Brand")

# Horizontal axis par price ka label diya
plt.xlabel("Average Price (INR)")

# Vertical axis par brand ka label diya
plt.ylabel("Brand")

# Gridlines lagayi taki price values esily track ho ske
plt.grid(axis ='x', linestyle='--',alpha =0.7)

# Margins fit karke final chart display kiya
plt.tight_layout()
plt.show()



```

```python
#====================================
#Que. RAM Capacity vs Average Price
#====================================
#soln. Ram ke according data group karke latest_price ka average calculate kiya
ram_avg_price = df.groupby('ram_gb')['latest_price'].mean().round(2)

#Numeric order mein Ram ko sort kiya taaki line chart smooth trend dikhaye
ram_avg_price = ram_avg_price.sort_index()

#Output values check karne ke liye print kiya
print(ram_avg_price)

#Chart ka canvas size set kiya
plt.figure(figsize=(8,4))

#line chart banaya marker (dots) ke sath naye visual look ke liye
plt.plot(ram_avg_price.index.astype(str)+ 'GB', ram_avg_price.values,
         color='#2ca02c', marker='o',linewidth=2.5, markersize=8)

#Chart ka title set kiya
plt.title("Impact of RAM Capacity on Laptop Price")

#Horizontal axis par RAM Capacity ka laber lagaya
plt.xlabel("RAM Capacity")

##Verical axis par price ka label lagaya
plt.ylabel("Average Price(INR)")

#Trend points clearly dekhne ke liye grid lagaya
plt.grid(True,linestyle=':',alpha=0.6)

#Canvas fit karke graph show kiya
plt.tight_layout()
plt.show()
```

```python
#========================================================
#que.4. Brand-wise Average Discount Percentage Analysis
#========================================================

# NOTE: Please make sure to run the data loading and cleaning cells first so 'df' and 'plt' are defined.
import matplotlib.pyplot as plt

#Agar Discount_percentage column mein '%' string jud chuka tha,
#toh calculation ke liye number banaya
if df['discount_percentage'].dtype == 'object':
  numeriac_discount = df['discount_percentage'].astype(str).str.rstrip('%').astype(float)
else:
  numeriac_discount = df['discount_percentage']

# Brand ke according group krke average discount calculate kiya aur sort kiya
avg_discount = numeriac_discount.groupby(df['brand']).mean().round(2).sort_values(ascending=False)

#Values screen par check krne ke liye print kiya
print(avg_discount)

#Chart frame ka size set kiya
plt.figure(figsize=(10,6))

#Purple color ka bar plot draw kiya
bars = plt.bar(avg_discount.index, avg_discount.values, color='#9467bd', edgecolor='black')

#har bar ke theek upar percentage value ka label lagaya taki exact value saaf dikhe
for bar in bars:
  yval = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.5, f'{yval:.1f}%', ha='center', va='bottom', fontsize=9)

#Title aur axis labels set kiye (Moved outside the loop so they only run once)
plt.title("Average Discount Percentage by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Discount (%)")
plt.xticks(rotation=45, ha='right')
plt.subplots_adjust(bottom=0.25, top=0.9)
plt.tight_layout()
plt.show()
```

```python
#=================================================
#Task 5. Processor Share Analysis (Donut Chart)
#=================================================

#Processor column ki total frequency count nikaali
processor_counts = df['processor_brand'].value_counts()

print(processor_counts)

#Canvas ka size set kiya
plt.figure(figsize=(6,6))

#Pie chart render kiya donut hole ke sath
plt.pie(processor_counts.values, labels=processor_counts.index,autopct='%1.f%%',
        startangle = 140,colors = ['#4C72B0','#55A868','#C44e52','#8172b3'],
        wedgeprops= dict(width=0.4,edgecolor='white'))

#Chart ka title set kiya
plt.title("Processor Brand Market Share")
plt.tight_layout()
plt.show()
```

```python
from matplotlib.lines import lineStyles
#Task 6. Top 5 Most Expensive Laptops (Ranked Horizontal Bar Chart)

#Dataset se top 5 highest latest_price wale records filter kiye
top5_expensive = df.nlargest(5,'latest_price')[['model','latest_price']].sort_values(by='latest_price',ascending=True)

#Output screen par check
print(top5_expensive)

#Canvas frame sizing set ki
plt.figure(figsize=(9,4))

#Crimson red color ka horizontal bar chart draw kiya
plt.barh(top5_expensive['model'],top5_expensive['latest_price'], color='#d62728',edgecolor = 'black')

#Visual title aur exes labels assign
plt.title("Top 5 Most Expensive Laptops in Market")
plt.xlabel("Price(INT)")
plt.ylabel("Laptop Model")
plt.grid(axis='x',linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
